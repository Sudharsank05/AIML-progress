```python
import pandas as pd
from scipy.stats import chi2_contingency
df = pd.read_csv('loan_approval.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df
```

```python
table = pd.crosstab(df['name'], df['loan_approved'])
print(table)

chi2, p, dof, expected = chi2_contingency(table)
alpha = 0.05
print(f'Chi2: {chi2:.2f}, p-value: {p:.4f}, Degrees of freedom: {dof}')
```

```python
df.drop(columns='name', inplace=True)
```

```python
table = pd.crosstab(df['city'], df['loan_approved'])
print(table)

chi2, p, dof, expected = chi2_contingency(table)
alpha = 0.05
print(f'Chi2: {chi2:.2f}, p-value: {p:.4f}, Degrees of freedom: {dof}')
```

```python
if p < alpha:
    print('relation')
else:
    print('No relation')
```
