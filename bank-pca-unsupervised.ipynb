```python
input = df.drop(columns=["Bankrupt?", "Net Income Flag"], errors="ignore")

# Keep numeric feature columns only
input = input.select_dtypes(include=np.number)

print("Null Values in 'input':", input.isnull().sum().sum())

# Fill missing values, if any
# input = input.fillna(input.median())

# Remove constant columns, if any
input = input.loc[:, input.nunique() > 1]

print("Features used:", input.shape[1])
```

```python
from sklearn.preprocessing import StandardScaler

st = StandardScaler()
x_scaled = st.fit_transform(input)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=40)
x_pca = pca.fit_transform(x_scaled)

x_pca
```

```python
print("egenvalues")  # variance
print(pca.explained_variance_)
print("eigenvectors")   #directions
print(pca.components_)
```
