```python
import pandas as pd
from scipy.stats import chi2_contingency

# Load your file
df = pd.read_csv('hotel_bookings_updated_2024.csv')
df.head()
```

```python
table = pd.crosstab(df['country'], df['is_canceled'])
print(table)

chi2, p, dof, expected = chi2_contingency(table)
alpha = 0.05
print(f'Chi2: {chi2:.2f}, p-value: {p:.4f}, Degrees of freedom: {dof}')
```

```python
if p < alpha:
    print('relation')
else:
    print('No relation')
```
