```python
categorical_cols = ['job', 'marital', 'education', 'default', 'housing', 'loan', 'contact', 'month', 'poutcome']
numerical_cols = ['age', 'balance', 'day', 'duration', 'campaign', 'pdays', 'previous']
target = 'deposit'

print("Categorical:", categorical_cols)
print("Numerical:", numerical_cols)
print("Target:", target)
```

```python
# Chi-Square Test for categorical features
from scipy.stats import chi2_contingency

print("=" * 70)
print("CHI-SQUARE TEST (Categorical vs Defaulted)")
print("=" * 70)

chi2_results = {}

for col in categorical_cols:
    table = pd.crosstab(df[col], df[target])
    chi2, p, dof, expected = chi2_contingency(table)

    chi2_results[col] = {'Chi2': chi2, 'p-value': p, 'dof': dof}

    print(f"\n{col} vs {target}")
    display(f"  Chi2: {chi2:.4f}, p-value: {p:.4f}, dof: {dof}")

    if p < 0.05:
        print(f"  ✅ Significant association! (p < 0.05)")
    else:
        print(f"  ❌ No significant association (p >= 0.05)")

# Summary
chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values('p-value')
print("\n" + "=" * 70)
print("CHI-SQUARE SUMMARY")
print("=" * 70)

pd.set_option('display.max_rows', None)
display(chi2_df)

```

```python
print("\n" + "=" * 70)
print("ANOVA TEST (Numerical vs Churn)")
print("=" * 70)

anova_results = {}

for col in numerical_cols:
    formula = f'{col} ~ C({target})'
    model = ols(formula, df).fit()
    aov_table = sm.stats.anova_lm(model, typ=2)

    f_value = aov_table['F'].iloc[0]
    p_value = aov_table['PR(>F)'].iloc[0]

    anova_results[col] = {'F-value': f_value, 'p-value': p_value}

    print(f"\n{col} vs {target}")
    display(f"  F-value: {f_value:.4f}, p-value: {p_value:.4f}")

    if p_value < 0.05:
        print(f"  ✅ Significant difference! (p < 0.05)")
    else:
        print(f"  ❌ No significant difference (p >= 0.05)")

# Summary
anova_df = pd.DataFrame(anova_results).T
anova_df = anova_df.sort_values('p-value')
print("\n" + "=" * 70)
print("ANOVA SUMMARY")
print("=" * 70)
display(anova_df)
```

```python
x = df.drop('deposit', axis=1)
y = df['deposit']

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
print("X_train shape:", x_train.shape)
print("X_test shape:", x_test.shape)
```

```python
# without hyperparamter tuning and with hyperparameter tuning on RandomForest Algorithm
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

rf = RandomForestClassifier(
    n_estimators=500,
    max_depth=5,
    min_samples_split=5,
    min_samples_leaf=2,
    class_weight="balanced"
)

rf.fit(x_train, y_train)

y_train_predict = rf.predict(x_train)

# Evaluation - train
print("Random Forest accuracy:", accuracy_score(y_train, y_train_predict))
print("Confusion matrix:\n", confusion_matrix(y_train, y_train_predict))
print("Classification report:\n", classification_report(y_train, y_train_predict))
```
