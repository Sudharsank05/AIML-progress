```python
import pandas as pd
from scipy.stats import chi2_contingency

# Load your file
df = pd.read_csv('hotel_bookings_updated_2024.csv')
df.head()
```

```python
table = pd.crosstab(df['country'], df['is_canceled'])
print(table)

chi2, p, dof, expected = chi2_contingency(table)
alpha = 0.05
print(f'Chi2: {chi2:.2f}, p-value: {p:.4f}, Degrees of freedom: {dof}')
```

```python
if p < alpha:
    print('relation')
else:
    print('No relation')
```

```python
import pandas as pd
from statsmodels.formula.api import ols
df = pd.read_csv('hotel_bookings_updated_2024.csv')
df.columns
```

```python
df
```

```python
df['required_car_parking_spaces'].unique()
```

```python
from statsmodels.stats.anova import _get_covariance, anova_lm
formula = 'is_canceled ~ lead_time + arrival_date_year + arrival_date_week_number + arrival_date_day_of_month + stays_in_weekend_nights + stays_in_week_nights + adults + days_in_waiting_list + adr + required_car_parking_spaces + total_of_special_requests'
model = ols(formula, df).fit()
aov_table = anova_lm(model)
print(aov_table)
```
