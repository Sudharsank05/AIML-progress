# PRINCIPAL COMPONENT ANALYSIS (PCA)

## 1. Introduction

PCA stands for **Principal Component Analysis**.

PCA is an unsupervised machine learning technique used for **dimensionality reduction**.

Dimensionality reduction means reducing the number of features (columns) in a dataset while keeping as much important information as possible.

The main idea of PCA is to transform a dataset with many features into a smaller set of new features called Principal Components.

**Example:**

Imagine a student dataset containing the following details:

- Mathematics Marks
- Science Marks
- English Marks
- Computer Science Marks
- History Marks

If we have many features, analyzing the data becomes difficult.

PCA helps reduce the number of features while preserving as much useful information as possible.

This makes the dataset easier to analyze and visualize.

## 2. What is Dimensionality Reduction?

Dimensionality Reduction is the process of reducing the number of input features in a dataset.

It helps simplify complex datasets without losing too much important information.

**Example:**

Consider a dataset containing 10 features.

After applying PCA, we may reduce these 10 features into 2 Principal Components.

- Original Features = 10
- Reduced Features = 2

This makes the dataset easier to understand and process.

## 3. What are Principal Components?

Principal Components are new variables created by combining the original features of a dataset.

They capture the maximum possible variation (information) present in the data.

There are two important points:

- First Principal Component (PC1): Captures the maximum variance in the dataset.
- Second Principal Component (PC2): Captures the next highest variance and is perpendicular (uncorrelated) to PC1.

**Example:**

If a dataset contains 5 original features, PCA creates new components such as:

- PC1
- PC2
- PC3
- PC4
- PC5

We can select only the first few components that preserve enough information.

## 4. How Does PCA Work?

Let's understand the working process step by step.

1. Collect Data: Select the dataset containing multiple features.
2. Standardize the Data: Bring all features to a similar scale so that features with larger values do not dominate the analysis.
3. Calculate Covariance Matrix: Identify how different features are related to each other.
4. Calculate Eigenvalues and Eigenvectors: Find the important directions and the amount of variance captured in each direction.
5. Rank Principal Components: Arrange the components based on their eigenvalues, from highest to lowest.
6. Select Components: Choose the most important Principal Components.
7. Transform the Data: Convert the original dataset into a reduced dataset using the selected components.

## 5. What is Variance in PCA?

Variance measures how much the data values are spread out from their average.

In PCA, variance helps identify which directions contain the most information.

- High Variance: Contains more variation in the data.
- Low Variance: Contains less variation in the data.

PCA selects directions with higher variance to preserve important information.

## 6. Eigenvalues and Eigenvectors

PCA uses Eigenvalues and Eigenvectors to identify the important Principal Components.

### 6.1 Eigenvalues

Eigenvalues represent the amount of variance captured by each Principal Component.

A higher eigenvalue means that the component captures more variance.

**Example:**

| Principal Component | Eigenvalue |
|---|---:|
| PC1 | 8.5 |
| PC2 | 4.2 |
| PC3 | 1.5 |
| PC4 | 0.8 |

Here, PC1 captures the highest variance, followed by PC2.

### 6.2 Eigenvectors

Eigenvectors represent the directions of the Principal Components.

They help identify how the original features contribute to each Principal Component.

**In Simple Words:**

- Eigenvalues = How much information (variance) is captured.
- Eigenvectors = Direction in which the information is captured.

## 7. Explained Variance Ratio

Explained Variance Ratio tells us how much of the total variance is captured by each Principal Component.

**Formula:**

Explained Variance Ratio = Eigenvalue of a Component / Sum of All Eigenvalues

**Example:**

Suppose the total variance is 10.

- PC1 captures 8.
- PC2 captures 2.

Therefore:

- PC1 = 80%
- PC2 = 20%

If we select PC1 and PC2, they preserve 100% of the variance in this example.

The explained variance ratio helps us decide how many Principal Components to select.

## 8. Advantages of PCA

1. Reduces the number of features in a dataset.
2. Helps reduce computational complexity.
3. Removes redundant information caused by correlated features.
4. Makes high-dimensional data easier to visualize.
5. Can help reduce noise in some datasets.

## 9. Disadvantages of PCA

1. Some information may be lost during dimensionality reduction.
2. Principal Components are difficult to interpret.
3. Data standardization is important when features have different scales.
4. PCA works mainly with linear relationships.
5. Selecting the correct number of components can be challenging.

## 10. Real-Time Applications

- Image Compression: Reduces the number of features in image data.
- Face Recognition: Reduces image dimensions while retaining important patterns.
- Finance: Identifies important patterns in financial data.
- Healthcare: Reduces the number of features in medical datasets.
- Data Visualization: Converts high-dimensional data into 2D or 3D data.
- Machine Learning: Reduces the number of input features before training a model.

## 11. Difference Between PCA and Feature Selection

| PCA | Feature Selection |
|---|---|
| Creates new features from original features. | Selects important features from existing features. |
| Combines multiple original features. | Keeps selected original features. |
| New features may be difficult to interpret. | Selected features are easier to interpret. |
| Reduces dimensionality by transformation. | Reduces dimensionality by removing less important features. |
| Example: PC1 and PC2. | Example: Age and Salary. |

## 12. Conclusion

Principal Component Analysis (PCA) is an unsupervised machine learning technique used for dimensionality reduction. It converts a large number of original features into a smaller number of Principal Components while preserving as much variance as possible.

PCA uses Eigenvalues and Eigenvectors to identify important directions in the dataset.

It is widely used in image processing, data visualization, healthcare, finance, and machine learning.

**Important Points for Students:**

- PCA = Principal Component Analysis.
- Main Purpose = Dimensionality Reduction.
- PC1 = Captures the maximum variance.
- PC2 = Captures the next highest variance.
- Eigenvalues = Amount of variance captured.
- Eigenvectors = Direction of Principal Components.
- Explained Variance Ratio = Percentage of variance captured by a component.
- Standardization = Brings features to a similar scale.
