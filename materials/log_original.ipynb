```python
df['Cloud9am']=df['Cloud9am'].fillna(df['Cloud9am'].median())
df['Cloud3pm']=df['Cloud3pm'].fillna(df['Cloud3pm'].median())
df['Temp3pm']=df['Temp3pm'].fillna(df['Temp3pm'].median())
df['Temp9am']=df['Temp9am'].fillna(df['Temp9am'].median())
#df['RainTomorrow']=df['RainTomorrow'].fillna(df['RainTomorrow'].mode()[0])
df['Temp9am']=df['Temp9am'].fillna(df['Temp9am'].median())
df.isnull().sum()
#df.dropna(inplace=True)
```

```python
x=df.drop(["RainToday"],axis=1)


y=df["RainToday"]
x
#y
```

```python
y
```

```python
from sklearn.preprocessing import MinMaxScaler  #StandardScaler
sc=MinMaxScaler()
c=sc.fit_transform(x)
print(c)
#
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
from sklearn.linear_model import LogisticRegression  #LinearRegression
lr=LogisticRegression()
lr.fit(x_train,y_train)
y_pred=lr.predict(x_train)
y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report  #r2_score,mean_squared_error,mean_absolute_error
print(accuracy_score(y_train,y_pred))
print(confusion_matrix(y_train,y_pred))
print(classification_report(y_train,y_pred))
#
```
