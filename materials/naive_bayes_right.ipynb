Here is a *complete guide to Naive Bayes models*, covering the theory, types, assumptions, formulas, implementation, advantages, and limitations.

---

## 🔍 What is Naive Bayes?

*Naive Bayes* is a probabilistic classification algorithm based on *Bayes’ Theorem, with the "naive" assumption that features are **conditionally independent* given the class label.

image.png

---

## 📘 Bayes’ Theorem

$$
P(Y|X) = \frac{P(X|Y) \cdot P(Y)}{P(X)}
$$
bayes theorem describes how to update the probability of a hypothesis based on new evidence


Where:

* $P(Y|X)$: Posterior probability of class $Y$ given features $X$
* $P(X|Y)$: Likelihood of features $X$ given class $Y$
* $P(Y)$: Prior probability of class $Y$
* $P(X)$: Marginal likelihood of features $X$

In practice:

$$
\text{Predict class } Y = \arg\max_{y} \; P(Y=y) \prod_{i=1}^{n} P(X_i | Y=y)
$$

---

## 🧠 Key Assumption

Naive Bayes *assumes feature independence*:

$$
P(X_1, X_2, ..., X_n | Y) = \prod_{i=1}^{n} P(X_i | Y)
$$

This is often *not true* in reality, but the model *still performs surprisingly well* in many practical scenarios.

---

## 🧪 Types of Naive Bayes Models

| Type               | Suitable for                   | Feature Assumption                             |
| ------------------ | ------------------------------ | ---------------------------------------------- |
| *Gaussian NB*    | Continuous data                | Features follow normal (Gaussian) distribution |
| *Multinomial NB* | Count data (e.g., word counts) | Features follow multinomial distribution       |
| *Bernoulli NB*   | Binary/boolean features        | Features are binary (0 or 1)                   |
| *Complement NB*  | Imbalanced text classification | Variant of Multinomial NB                      |

---

## ⚙ Steps in Naive Bayes Classification

1. *Calculate Priors*:

   $$
   P(Y=c) = \frac{\text{Count of class }c}{\text{Total samples}}
   $$

2. *Calculate Likelihoods*:

   * For *Gaussian*:

     $$
     P(X_i | Y=c) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right)
     $$
   * For *Multinomial/Bernoulli*: Use frequency of terms or binary presence.

3. *Use Bayes' theorem* to compute posterior for each class:

   $$
   \text{Posterior} \propto P(Y) \prod P(X_i | Y)
   $$

4. *Choose the class with the highest posterior*.

---

## ✅ Advantages

* Simple and fast to train
* Performs well on *small datasets*
* Handles *high-dimensional data* well (like text classification)
* Not sensitive to irrelevant features
* Works well with *categorical input features*

---

## ❌ Limitations

* Assumes *independence* between features (often unrealistic)
* Poor estimates for features with *zero frequency* (solved with *Laplace smoothing*)
* Doesn’t perform well if features are *highly correlated*
* Struggles with *continuous* variables unless Gaussian distribution is valid

---

## 🛠 Python Example: Gaussian Naive Bayes

python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, accuracy_score

# Load data
X, y = load_iris(return_X_y=True)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

# Model
model = GaussianNB()
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))


---

## 🧪 Laplace Smoothing (for Multinomial/Bernoulli)

To handle zero frequencies:

$$
P(x_i | y) = \frac{count(x_i, y) + \alpha}{count(y) + \alpha \cdot n}
$$

Where $\alpha = 1$ is the default Laplace smoothing value, and $n$ is number of features.
