## **1. What is CNN?**

**CNN stands for Convolutional Neural Network.**

CNN is a type of Deep Learning algorithm mainly used to identify and recognize images.

It helps computers understand the content of an image, just like humans recognize objects using their eyes and brain.

**Simple example:**

Imagine you show a computer two images:

- Image 1 – Cat 🐱
- Image 2 – Dog 🐶

CNN learns the features of cats and dogs, such as ears, eyes, nose, face shape and body structure. Then it predicts whether the image contains a cat or a dog.

## **4. Main Layers of CNN**

CNN has four important components.

### **A. Convolutional Layer**

- It is the first major layer in CNN.
- It extracts important features from an image.
- It uses a small matrix called a **Filter or Kernel**.
- It detects edges, lines, corners and patterns.

**Example:** When CNN receives a cat image, the convolution layer identifies the cat's eyes, ears and other features.

### **B. ReLU Activation Function**

ReLU stands for **Rectified Linear Unit**.

It is an activation function used to introduce non-linearity into the neural network.

Formula:

\\[ f(x)=\max(0,x) \\]

Simple explanation:

- If the input is negative, output is 0.
- If the input is positive, output remains the same.

| Input | Output |
| ----- | ------ |
| -5    | 0      |
| -2    | 0      |
| 3     | 3      |
| 8     | 8      |

### **C. Pooling Layer**

Pooling reduces the size of the feature map.

It helps reduce computation and keeps important information.

Two common types:

1. **Max Pooling:** Selects the maximum value.
2. **Average Pooling:** Calculates the average value.

## **6. Advantages of CNN**

1. Automatically extracts image features.
2. Provides good performance in image recognition tasks.

5. Needs careful tuning of parameters.

## **9. CNN vs Traditional Neural Network**

| CNN                                             | Traditional ANN                                    |
| ----------------------------------------------- | -------------------------------------------------- |
| Mainly used for image-related tasks             | Used for many structured-data tasks                |
| Uses convolution and pooling                    | Uses fully connected layers                        |
| Automatically extracts spatial features         | Often needs prepared input features                |
| Preserves spatial relationships in feature maps | Flattening image inputs can lose spatial structure |

## **10. Conclusion**

CNN is a powerful Deep Learning algorithm mainly used for image recognition and computer vision. It automatically extracts important features from images using convolution and pooling layers. Finally, it uses fully connected layers to classify the image.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Conv2D, MaxPooling2D,Flatten  #Fully connected layer,Reduces image size,Converts 2D matrix into 1D vector
import numpy as np
```

```python
x_train=np.loadtxt('/content/cat dog.csv',delimiter=',')
y_train=np.loadtxt('/content/cat dog 2.csv',delimiter=',')
x_test=np.loadtxt('/content/cat dog.csv',delimiter=',')
y_test=np.loadtxt('/content/cat dog 2.csv',delimiter=',')
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
#reshape
x_train=x_train.reshape(len(x_train),100,100,3)   # 3 RGB CHANNEL
y_train=y_train.reshape(len(y_train),1)
x_test=x_test.reshape(len(x_test),100,100,3)
y_test=y_test.reshape(len(y_test),1) #1 D COLUMNS CONVERT TO 2D
```

```python
x_train[1,:]   #X TRAIN IMAGES Y_TRAIN LABELS(0,1)   # 1,2,3,4,5
```

```python
#scaler
x_train=x_train/255
x_test=x_test/255       #255/255=1,128/255=0.5
```

```python
import matplotlib.pyplot as plt
plt.imshow(x_train[2])
```

```python
#model
#sequential ---conv,maxpooling,conv,maxpooling................flatten(dense)  #cnn build
model=Sequential([
    Conv2D(32,(3,3),activation='relu',input_shape=(100,100,3)),#filters ,height & width
    MaxPooling2D((2,2)),#filter size  ,100*100=50*50
    Conv2D(32,(3,3),activation='relu'),
    MaxPooling2D((2,2)),
    Flatten(), # cnn output convert to ann
    Dense(64,activation='relu'),# 64 is a neurons
    Dense(1,activation='sigmoid')
])
```

```python
model=Sequential()
model.add(Conv2D(32,(3,3),activation='relu',input_shape=(100,100,3)))
model.add(MaxPooling2D((2,2)))
model.add(Conv2D(32,(3,3),activation='relu'))
model.add(MaxPooling2D((2,2)))
model.add(Flatten())
model.add(Dense(64,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
```

```python
# cost function & back propagation
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])  #training rules set model ,AdaGrad, RMSProp, Momentum
#optimizer= model weight update,error remove,backpropagaation use to improve accurcy, adam is fast
#loss= to measure prediction correct or wrong
#accuracy during the train to find the accuracy
```

```python
model.fit(x_test,y_test,epochs=5,batch_size=64)  #batch 640 images,640/64
```

```python
model.evaluate(x_test,y_test)
```
