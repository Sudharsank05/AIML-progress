# DEEP LEARNING

## 1. Introduction to Deep Learning

Deep Learning is a branch of Artificial Intelligence (AI) and Machine Learning (ML).

It helps computers learn from large amounts of data and make decisions like humans.

Deep Learning works using a structure called an **Artificial Neural Network (ANN)**, which is inspired by the human brain.

**Simple Example:**

Imagine a child sees pictures of cats and dogs.

Initially, the child may not know the difference between a cat and a dog.

But after seeing many pictures:
- The child learns that cats have certain features.
- The child learns that dogs have different features.
- Finally, the child can identify whether a new picture contains a cat or a dog.

Similarly, Deep Learning trains computers using large amounts of data to recognize patterns and make predictions.

## 2. What is Artificial Intelligence (AI)?

Artificial Intelligence is a technology that enables computers and machines to perform tasks that normally require human intelligence.

These tasks include:
- Learning
- Problem-solving
- Decision-making
- Understanding language
- Recognizing images

**Examples:**
- Google Assistant
- ChatGPT
- Self-driving cars
- Face recognition systems

## 3. What is Machine Learning (ML)?

Machine Learning is a branch of Artificial Intelligence that allows computers to learn from data without being explicitly programmed for every task.

Instead of writing rules for every situation, we provide data to the computer so that it can learn patterns.

**Example:**

Imagine we provide a computer with information about houses:

- House size
- Number of bedrooms
- Location
- House price

The machine learns the relationship between these features and house prices.

After training, it can predict the price of a new house.

## 4. What is Deep Learning?

Deep Learning is a subset of Machine Learning that uses multiple layers of artificial neural networks to learn complex patterns from data.

It can automatically learn important features from images, audio, text and other types of data.

**Examples:**
- Face recognition
- Voice assistants
- Language translation
- Self-driving cars
- Medical image analysis

## 5. Difference Between AI, ML and Deep Learning

**Remember:**

Deep Learning is inside Machine Learning, and Machine Learning is inside Artificial Intelligence.

## 6. What is a Neural Network?

A Neural Network is a machine learning model inspired by the way the human brain works.

The human brain contains neurons that communicate with each other.

Similarly, an artificial neural network contains artificial neurons that process information and learn patterns from data.

**Simple Example:**

Imagine we want a computer to identify whether an image contains a cat.

1. We provide a cat image to the computer.
2. The neural network processes the image.
3. Different neurons identify patterns such as edges, shapes and features.
4. The network combines the learned information.
5. It predicts that the image contains a cat.

## 7. Structure of an Artificial Neural Network (ANN)

An Artificial Neural Network generally contains three main types of layers.

### 7.1 Input Layer

The Input Layer receives data from the user or dataset.

**Example:**

For house price prediction, the input features may include:
- House size
- Number of bedrooms
- Location

### 7.2 Hidden Layer

The Hidden Layer processes the input data and learns important patterns.

A neural network can contain one or multiple hidden layers.

Deep Learning models generally use multiple hidden layers.

### 7.3 Output Layer

The Output Layer provides the final prediction or result.

**Example:**

For house price prediction, the output may be:

Predicted House Price = ₹50,00,000

**Simple Flow:**

Input Layer → Hidden Layers → Output Layer

## 8. How Does Deep Learning Work?

Let's understand the working process step by step.

1. Collect Data: Gather the required data, such as images, text, audio or numbers.
2. Prepare Data: Clean and organize the data before training.
3. Build a Neural Network: Create a model containing input, hidden and output layers.
4. Train the Model: Provide training data to the neural network so it can learn patterns.
5. Calculate Error: Compare the predicted result with the actual result.
6. Update Weights: Adjust the model's internal weights to reduce prediction errors.
7. Repeat Training: Continue the process for multiple iterations until the model learns useful patterns.
8. Make Predictions: Use the trained model to predict results for new data.

## 9. Important Concepts in Deep Learning

### 9.1 Neuron

A Neuron is the basic processing unit of an artificial neural network.

It receives inputs, processes them and produces an output.

### 9.2 Weights

Weights determine how important each input is to the model's prediction.

The model learns suitable weight values during training.

### 9.3 Bias

Bias is an additional value used by a neuron to adjust its output.

It helps the model learn more flexible patterns.

### 9.4 Activation Function

An Activation Function determines the output of a neuron.

It helps neural networks learn complex, non-linear relationships.

Common activation functions:
- ReLU
- Sigmoid
- Tanh
- Softmax

### 9.5 Epoch

An Epoch means one complete pass through the entire training dataset.

**Example:**

If a dataset contains 1,000 images and the model processes all 1,000 images once, that is one epoch.

### 9.6 Loss Function

A Loss Function measures the difference between the actual output and the predicted output.

The goal of training is to reduce the loss.

### 9.7 Backpropagation

Backpropagation is a learning process used to calculate how the model's weights should be adjusted to reduce errors.

It sends error information backward through the network.

### 9.8 Optimizer

An Optimizer updates the model's weights using the calculated gradients to reduce the loss.

Examples:
- SGD
- Adam
- RMSprop

## 10. Types of Deep Learning Networks

There are different types of neural network architectures.

### 10.1 Artificial Neural Network (ANN)

ANN is a basic neural network architecture that contains input, hidden and output layers.

**Applications:**
- House price prediction
- Customer classification
- Sales prediction

### 10.2 Convolutional Neural Network (CNN)

CNN is mainly used for image and visual data processing.

It automatically identifies features such as edges, shapes and objects in images.

**Applications:**
- Face recognition
- Object detection
- Medical image analysis
- Image classification

### 10.3 Recurrent Neural Network (RNN)

RNN is designed to process sequential data.

It uses information from previous steps to help process the current step.

**Applications:**
- Text processing
- Speech recognition
- Time series forecasting

### 10.4 Long Short-Term Memory (LSTM)

LSTM is a special type of Recurrent Neural Network.

It is designed to remember important information over longer sequences.

**Applications:**
- Language modelling
- Speech recognition
- Time series prediction

### 10.5 Transformer

Transformers are deep learning architectures that use an attention mechanism to understand relationships between different parts of input data.

They are widely used in modern language models.

**Applications:**
- ChatGPT
- Language translation
- Text generation
- Question answering

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split1
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
df=pd.read_csv("/content/pizza_sales.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df=df.drop(columns=['pizza_id','order_id','pizza_name_id'])
```

```python
df.info()
```

```python
df.select_dtypes(include=['object'])
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed', dayfirst=True)
df['order_time'] = pd.to_datetime(df['order_time'], format='%H:%M:%S').dt.time
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
for col in df.select_dtypes(include=['object']).columns:
    df[col]=le.fit_transform(df[col])
```

```python
df.info()
```

```python
x=df.drop(['total_price'],axis=1)
y=df['total_price']
```

```python
x
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=47)
```

```python
scaler=StandardScaler()
x_train['order_date'] = x_train['order_date'].astype('int64') // 10**9
x_test['order_date'] = x_test['order_date'].astype('int64') // 10**9
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
x_train.shape[1]
```

```python
model=Sequential([
    Dense(500,activation='relu',input_shape=(x_train.shape[1],)),
    Dense(300,activation='relu'),
    Dense(150,activation='relu'),
    Dense(1)
])
```

```python
model.compile(optimizer='adam',loss='mae',metrics=['mae'])
```

```python
history=model.fit(x_train,y_train,epochs=10,validation_data=(x_test,y_test))
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['mae'])
plt.plot(history.history['val_mae'])
plt.title('model mae')
plt.ylabel('mae')
plt.xlabel('epoch')
plt.legend(['train','test'],loc='upper left')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('model mae')
plt.ylabel('mae')
plt.xlabel('epoch')
plt.legend(['train','test'],loc='upper left')
plt.show()
```
