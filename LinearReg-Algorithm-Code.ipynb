```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# data
x = np.array([[10], [15], [20], [25], [30], [35]])
y = np.array([100, 150, 200, 250, 300, 350])

# model
model = LinearRegression()

# train model
model.fit(x, y)

# coeffecients
# slope
print("Slope:", {model.coef_[0]})
# intercept
print("intercept: ", model.intercept_)

# test data
x_test = np.array([[40], [45], [50]])
y_test = np.array([400, 450, 500]) #actual values

# prediction
prediction = model.predict(x_test)
print("Predictions:")
for i, x in enumerate(x_test):
    print(f"{x[0]} -> {prediction[i]:.2f}   (Actual: {y_test[i]})")

# evaluate
mse = mean_squared_error(y_test, prediction)
r2 = r2_score(y_test, prediction)
print(f"\nMSE: {mse:.2f}")
print(f"R2: {r2:.4f}")


# visualize
x = np.array([[10], [15], [20], [25], [30], [35]])
y = np.array([100, 150, 200, 250, 300, 350])

plt.scatter(x, y, color='blue', label='training data')
plt.plot(x, model.predict(x), color='red', label='regression line')
plt.scatter(x_test, y_test, color='green', label='test data')
plt.scatter(x_test, prediction, color='orange', label='predicted output', marker='x')
plt.title("Linear Regression")
plt.legend()
plt.show()
```
