```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
#normalizing
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(rescale=1./255)
```

```python
#generator
train_data = train_datagen.flow_from_directory(
    r"/Users/sudharsank/AIML/deep-learning/CNN/melanoma-prediction-project/melanoma_cancer_dataset/train",
    target_size=(224,224),
    batch_size=32,
    class_mode='binary'
)

val_data = val_datagen.flow_from_directory(
    r"/Users/sudharsank/AIML/deep-learning/CNN/melanoma-prediction-project/melanoma_cancer_dataset/test",
    target_size=(224,224),
    batch_size=32,
    class_mode='binary'
)

print("Training Classes: ", train_data.class_indices)
```
